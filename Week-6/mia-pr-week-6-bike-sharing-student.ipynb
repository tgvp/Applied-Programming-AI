## Cheatsheet — Generic patterns from earlier classes

These are reference patterns, not a sequence to execute or a solution to the activity. Names such as `measurement`, `category` and `x_column` are illustrative. Choose the columns, conditions and methods that fit your analysis. All examples assume a DataFrame called `df` and the imports from the setup cell.

### Inspect a table
```python
df.head()
df.shape
df.info()
df.describe(include='all')
df.columns
df.dtypes
```

### Select, drop and rename columns
```python
df[['column_a', 'column_b']]
df = df.drop(columns=['unwanted_column'])
df = df.rename(columns={'old_name': 'new_name'})
```

### Inspect and remove duplicate rows
```python
df.duplicated().sum()
df = df.drop_duplicates()
```

### Standardise text categories
```python
df['category'].value_counts(dropna=False)
df['category'] = df['category'].astype('string').str.strip().str.lower()
```

### Remove literal unit text and convert numbers
```python
text = df['measurement'].astype('string')
text = text.str.replace(' kg', '', regex=False).str.strip()
df['measurement'] = pd.to_numeric(text, errors='coerce')
```

`errors='coerce'` turns text that cannot be parsed into missing values, not zero. Choose the literal unit string that actually appears in your data.

### Inspect values using a condition
```python
condition = df['measurement'] < 0
condition.sum()
df.loc[condition, ['identifier', 'measurement']]
df.loc[condition, 'measurement'] = np.nan
```

The condition is only an example. Decide which values are impossible using the meaning and units of the variable.

### Inspect missingness and choose a policy
```python
df.isna().sum()
df.isna().mean() * 100

# Possible choices, not instructions to apply both:
filtered = df.dropna(subset=['required_column'])
filled = df['measurement'].fillna(df['measurement'].median())
```

Filtering or filling can change an analysis. Justify your choice and report how many observations it affects. Save the result to your working table if you decide to use it.

### Filter with two conditions
```python
selected = df[(df['x_column'] >= 5) & (df['y_column'] < 10)]
selected[['category', 'x_column', 'y_column']]

```

Use parentheses around each condition. `&` means both conditions must hold; `|` means at least one must hold.
