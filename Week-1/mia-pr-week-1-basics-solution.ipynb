### For Statements

```python
for <var> in <iterable>:
    <statement(s)>
```

````python
#for loop will interact with an object and return all parts of that object
#using the structure above, create a for loop that iterates over the list [1,2,3,5] and print the values

for number in [1, 2, 3, 5]:
    print(number)
````

#### range is often used as the iterable

```python
    for <var> in range(limit):
        <statement(s)>
```

````python
#We can even use a range and manipulate the addresses of a list
#using the len() function, create a for loop with range that prints the addresses of the list [1,2,3,5]

numbers = [1, 2, 3, 5]
for index in range(len(numbers)):
    print(index)
````

### If Statements

```python
    if <expr>:
        <statement>
    elif <expr>:
        <statement>
    else:
        <statement>
```

`expr` is an expression evaluated in a Boolean context
    
`statement` is a valid Python statement, which must be indented

````python
#If in Python works just like in other languages but without the need for '()' or '{}'
#It is mandatory to indent the if structure for it to work
#Create an if statement that tests if 2 > 4. If true, print 'yes', otherwise print 'no'

if 2 > 4:
    print("yes")
else:
    print("no")
````

### List Comprehensions

Create lists with elements with a single statement

```python
[ expression for a_value in a_collection ]

```

````python
#create a loop structure that iterates over the list [1,2,3,4,5] and prints the values +10 Ex: [11,12,13,14,15]
#now using list comprehensions, create the same structure as above

numbers = [1, 2, 3, 4, 5]
shifted_numbers = []
for number in numbers:
    shifted_numbers.append(number + 10)
print(shifted_numbers)

# The same result using a list comprehension.
shifted_numbers = [number + 10 for number in numbers]
print(shifted_numbers)
````

List comprehension with conditions

```python
[ expression for a_value in a_collection if condition ]

```

````python
#Following the same structure above, now print the number plus 10 for only the values greater than 3 in the list [1, 2, 3, 4, 5].

numbers = [1, 2, 3, 4, 5]
print([number + 10 for number in numbers if number > 3])
````

#### Exercise 12 - Use a list, a loop, and an if in a function

Complete `keep_above(numbers, limit)` so it returns a new list containing only numbers greater than `limit`. Keep the numbers in their original order.

Reuse the idea from Exercise 8:
1. Visit each number with a `for` loop.
2. Use an `if` statement to compare it with the parameter `limit`.
3. Append it to `selected` only if it is greater than `limit`.

Keep `return selected` after the loop, inside the function.

**Expected output:**
```text
[8, 6]
[8]
[]
```

````python
def keep_above(numbers, limit):
    """Return a list of numbers greater than limit."""
    selected = []
    # TODO: Loop over numbers, check each one, and append when appropriate.
    for number in numbers:
        if number > limit:
            selected.append(number)
    return selected


print(keep_above([2, 8, 5, 6, 1], 5))
print(keep_above([2, 8, 5, 6, 1], 7))
print(keep_above([2, 8, 5, 6, 1], 10))
````
