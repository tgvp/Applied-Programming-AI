## 3. Dispersion
The range spans the extremes. The interquartile range (IQR) measures the width of the middle half. Variance and standard deviation also use distances from the mean. Standard deviation is expressed in the original units.

````python
print("Age range:", age.max() - age.min())
print("Age IQR:", age.quantile(0.75) - age.quantile(0.25))
print("Age variance:", age.var())
print("Age standard deviation:", age.std())
````

## 6. Does survival rise with family size?
A family size of one means the passenger had no recorded relatives aboard. Each row of the summary below represents one family size. The mean of the Boolean survival column gives the fraction who survived. Multiply by 100 to express it as a percentage.

````python
groups = df.groupby("family_size")["survived"]
summary = groups.size().to_frame("passengers")
summary["survival_pct"] = groups.mean() * 100
summary = summary.reset_index()
summary.round(1)
````

**Example**

>```python
>plt.scatter(df["x_column"], df["y_column"])
>plt.show()
>```

````python
plt.scatter(summary["family_size"], summary["survival_pct"])
plt.xlabel("Family size (including the passenger)")
plt.ylabel("Passengers who survived (%)")
plt.show()
````
